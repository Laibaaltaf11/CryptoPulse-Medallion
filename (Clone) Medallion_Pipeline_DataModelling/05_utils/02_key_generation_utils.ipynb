```python
# Databricks notebook source

# ==========================================================
# Utility Notebook
#
# Notebook:
# 02_key_generation_utils
#
# Purpose:
# Centralized utility functions for deterministic
# surrogate key generation.
#
# ==========================================================

from pyspark.sql import functions as F

# ==========================================================
# Coin Surrogate Key
#
# Input:
# coin_id
# effective_start_date
#
# Output:
# SHA256 surrogate key
# ==========================================================

def generate_coin_sk(df):

    return (

        df.withColumn(

            "coin_sk",

            F.sha2(

                F.concat_ws(

                    "||",

                    F.col("coin_id"),

                    F.col("effective_start_date").cast("string")

                ),

                256

            )

        )

    )

# ==========================================================
# Record Hash
#
# Used for SCD Type 2 change detection
# ==========================================================

def generate_record_hash(df):

    return (

        df.withColumn(

            "record_hash",

            F.sha2(

                F.concat_ws(

                    "||",

                    F.col("coin_id"),

                    F.col("name"),

                    F.col("symbol"),

                    F.col("image")

                ),

                256

            )

        )

    )

# ==========================================================
# Generic Composite Key
#
# Example:
#
# df = generate_key(
#     df,
#     ["coin_id","observation_ts"],
#     "fact_key"
# )
#
# ==========================================================

def generate_key(df, columns, key_name):

    return (

        df.withColumn(

            key_name,

            F.sha2(

                F.concat_ws(

                    "||",

                    *[F.col(c).cast("string") for c in columns]

                ),

                256

            )

        )

    )

# ==========================================================
# Batch Key
#
# Useful for ingestion tracking
# ==========================================================

def generate_batch_key(df):

    return (

        df.withColumn(

            "batch_key",

            F.sha2(

                F.concat_ws(

                    "||",

                    F.col("_batch_id"),

                    F.col("_ingested_at").cast("string")

                ),

                256

            )

        )

    )

# ==========================================================
# Observation Key
#
# One key per coin snapshot
# ==========================================================

def generate_observation_key(df):

    return (

        df.withColumn(

            "observation_key",

            F.sha2(

                F.concat_ws(

                    "||",

                    F.col("coin_id"),

                    F.col("observation_ts").cast("string")

                ),

                256

            )

        )

    )

print("Key Generation Utilities Loaded Successfully.")
```
