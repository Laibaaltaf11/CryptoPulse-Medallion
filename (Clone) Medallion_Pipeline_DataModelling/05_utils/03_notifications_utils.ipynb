```python
# Databricks notebook source

# ==========================================================
# Utility Notebook
#
# Notebook:
# 03_notification_utils
#
# Purpose:
# Utility functions for pipeline logging and notifications.
#
# ==========================================================

from pyspark.sql import Row
from pyspark.sql import functions as F
from datetime import datetime, UTC

CATALOG = "crypto_pipeline"

NOTIFICATION_TABLE = f"{CATALOG}.meta.pipeline_notifications"

spark.sql(f"USE CATALOG {CATALOG}")

# ==========================================================
# Create Notification Table
# ==========================================================

spark.sql(f"""

CREATE TABLE IF NOT EXISTS {NOTIFICATION_TABLE}

(

pipeline_name STRING,

status STRING,

message STRING,

notified_at TIMESTAMP

)

USING DELTA

""")

# ==========================================================
# Success Notification
# ==========================================================

def notify_success(pipeline_name, message):

    print(f"SUCCESS : {pipeline_name}")

    print(message)

    rows = [

        Row(

            pipeline_name=pipeline_name,

            status="SUCCESS",

            message=message,

            notified_at=datetime.now(UTC)

        )

    ]

    spark.createDataFrame(rows).write.mode("append").saveAsTable(NOTIFICATION_TABLE)

# ==========================================================
# Failure Notification
# ==========================================================

def notify_failure(pipeline_name, message):

    print(f"FAILED : {pipeline_name}")

    print(message)

    rows = [

        Row(

            pipeline_name=pipeline_name,

            status="FAILED",

            message=message,

            notified_at=datetime.now(UTC)

        )

    ]

    spark.createDataFrame(rows).write.mode("append").saveAsTable(NOTIFICATION_TABLE)

# ==========================================================
# Warning Notification
# ==========================================================

def notify_warning(pipeline_name, message):

    print(f"WARNING : {pipeline_name}")

    print(message)

    rows = [

        Row(

            pipeline_name=pipeline_name,

            status="WARNING",

            message=message,

            notified_at=datetime.now(UTC)

        )

    ]

    spark.createDataFrame(rows).write.mode("append").saveAsTable(NOTIFICATION_TABLE)

# ==========================================================
# Show Notification History
# ==========================================================

def show_notifications():

    display(

        spark.table(NOTIFICATION_TABLE)

        .orderBy(

            F.desc("notified_at")

        )

    )

print("Notification Utilities Loaded Successfully.")
```
